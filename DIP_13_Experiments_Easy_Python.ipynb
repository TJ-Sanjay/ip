## 1. Generate all possible 3 × 2 black-and-white images

**Idea:** Six pixels, each either black (0) or white (255), give 2⁶ = 64 images.

## 5. Read an image matrix and extract neighbourhoods

The 8-neighbourhood is the eight pixels around a centre pixel in a 3 × 3 window. Here, the 16-neighbourhood means the 16 perimeter pixels of a 5 × 5 window.

````python
img = np.array(Image.open("sample.jpg").convert("L"))
row, col = img.shape[0] // 2, img.shape[1] // 2

window3 = img[row-1:row+2, col-1:col+2]
print("Centre pixel:", img[row, col])
print("3x3 window:\n", window3)
print("8 neighbours:", np.delete(window3.flatten(), 4))

window5 = img[row-2:row+3, col-2:col+3]
neighbours16 = window5[[0,0,0,0,0,1,1,3,3,4,4,4,4,4,2,2],
                       [0,1,2,3,4,0,4,0,4,0,1,2,3,4,0,4]]
print("5x5 window:\n", window5)
print("16-neighbourhood:", neighbours16)
show(img, "Input image")
````

**Expected output:** The input image, selected centre pixel, 3 × 3 window, eight neighbours, and 16 perimeter pixels. Use an image at least 5 × 5 pixels.

## 8. Power-law (gamma) transformations of a selfie

Save your selfie as `selfie.jpg`. Gamma below 1 generally brightens the image; gamma above 1 generally darkens it.

````python
img = np.array(Image.open("selfie.jpg").convert("RGB"))
gammas = [0.5, 0.8, 1.0, 1.5, 2.0]

plt.figure(figsize=(15, 4))
for i, gamma in enumerate(gammas):
    result = (255 * (img / 255.0) ** gamma).astype(np.uint8)
    plt.subplot(1, 5, i+1)
    plt.imshow(result)
    plt.title("Gamma = " + str(gamma))
    plt.axis("off")
plt.show()
````

## 9. Convolution on repeated shapes

Choose the shape based on your registration-number rule. This example creates repeated circles and applies an edge-detection kernel.

````python
img = np.zeros((240, 240), dtype=np.uint8)
for x, y in [(60,60), (180,60), (60,180), (180,180)]:
    cv2.circle(img, (x,y), 25, 255, -1)

kernel = np.array([[-1,-1,-1], [-1,8,-1], [-1,-1,-1]], dtype=np.float32)
result = cv2.filter2D(img, -1, kernel)
result = np.clip(np.abs(result), 0, 255).astype(np.uint8)

plt.subplot(1, 2, 1); plt.imshow(img, cmap="gray"); plt.title("Input circles"); plt.axis("off")
plt.subplot(1, 2, 2); plt.imshow(result, cmap="gray"); plt.title("Convolution output"); plt.axis("off")
plt.show()
````

**Expected output:** Input shapes and their edge response. Change `cv2.circle` to `cv2.rectangle` for squares or use `cv2.fillPoly` for triangles.

## 10. Add Gaussian noise and remove it with a 5 × 5 averaging filter

This follows the specified Gaussian noise mean = 110 and standard deviation = 5. Since the mean is strongly positive, the image may become much brighter; that is an effect of the requested noise parameters.

````python
img = np.array(Image.open("selfie.jpg").convert("L")).astype(np.float32)
noise = np.random.normal(110, 5, img.shape)
noisy = np.clip(img + noise, 0, 255).astype(np.uint8)

kernel = np.ones((5, 5), dtype=np.float32) / 25
recovered = cv2.filter2D(noisy, -1, kernel)

mse_noisy = np.mean((img - noisy.astype(np.float32)) ** 2)
mse_recovered = np.mean((img - recovered.astype(np.float32)) ** 2)
print("MSE before filtering:", mse_noisy)
print("MSE after filtering:", mse_recovered)

plt.subplot(1, 3, 1); plt.imshow(img, cmap="gray", vmin=0, vmax=255); plt.title("Original"); plt.axis("off")
plt.subplot(1, 3, 2); plt.imshow(noisy, cmap="gray", vmin=0, vmax=255); plt.title("Noisy"); plt.axis("off")
plt.subplot(1, 3, 3); plt.imshow(recovered, cmap="gray", vmin=0, vmax=255); plt.title("Recovered"); plt.axis("off")
plt.show()
````

**Expected output:** Original, noisy, and recovered images plus MSE values. The recovered MSE may not be lower because averaging cannot remove a large positive noise bias.

## 11. Rotation, scaling, and shearing using explicit matrices

This code does not use ready-made image rotation, resizing, or affine-warp functions. It uses explicit coordinate matrices and manual inverse mapping with nearest-neighbour pixel selection.

````python
img = np.array(Image.open("selfie.jpg").convert("L"))
h, w = img.shape

def manual_transform(img, A):
    h, w = img.shape
    cx, cy = (w-1)/2, (h-1)/2
    T1 = np.array([[1,0,-cx], [0,1,-cy], [0,0,1]], dtype=float)
    T2 = np.array([[1,0,cx], [0,1,cy], [0,0,1]], dtype=float)
    M = T2 @ A @ T1
    invM = np.linalg.inv(M)
    out = np.zeros_like(img)

    for y in range(h):
        for x in range(w):
            sx, sy, _ = invM @ np.array([x, y, 1.0])
            sx, sy = int(round(sx)), int(round(sy))
            if 0 <= sx < w and 0 <= sy < h:
                out[y, x] = img[sy, sx]
    return out

theta = np.deg2rad(30)
R = np.array([[np.cos(theta), -np.sin(theta), 0],
              [np.sin(theta),  np.cos(theta), 0],
              [0, 0, 1]], dtype=float)
S = np.array([[1.3, 0, 0], [0, 1.3, 0], [0, 0, 1]], dtype=float)
H = np.array([[1, 0.4, 0], [0, 1, 0], [0, 0, 1]], dtype=float)

rotated = manual_transform(img, R)
scaled = manual_transform(img, S)
sheared = manual_transform(img, H)

plt.figure(figsize=(12, 4))
for i, (image, title) in enumerate([(img,"Original"),(rotated,"Rotation"),(scaled,"Scaling"),(sheared,"Shearing")]):
    plt.subplot(1,4,i+1); plt.imshow(image, cmap="gray"); plt.title(title); plt.axis("off")
plt.show()
````

**Challenges:** black empty areas, clipping at image borders, and jagged pixels due to nearest-neighbour sampling. The transformation matrices are used explicitly.

## 12. Track a leaf apex and apply FFT

Record a stable video of the leaf and save it as `leaf.mp4`. This simple tracker finds the topmost green pixel in each frame; it may need adjustment if the background also contains green.

````python
cap = cv2.VideoCapture("leaf.mp4")
rows, cols = [], []

while True:
    ok, frame = cap.read()
    if not ok:
        break
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv, np.array([25,40,30]), np.array([95,255,255]))
    y, x = np.where(mask > 0)

    if len(y) > 0:
        i = np.argmin(y)
        rows.append(y[i])
        cols.append(x[i])

cap.release()
rows = np.array(rows)
cols = np.array(cols)

if len(rows) > 1:
    plt.subplot(1,2,1); plt.plot(rows); plt.title("Row vs frame"); plt.xlabel("Frame")
    plt.subplot(1,2,2); plt.plot(cols); plt.title("Column vs frame"); plt.xlabel("Frame")
    plt.show()

    row_fft = np.fft.fft(rows - rows.mean())
    col_fft = np.fft.fft(cols - cols.mean())

    plt.subplot(1,2,1); plt.plot(np.abs(row_fft)); plt.title("1D FFT of row signal")
    plt.subplot(1,2,2); plt.plot(np.abs(col_fft)); plt.title("1D FFT of column signal")
    plt.show()

    signal_2d = np.vstack((rows, cols))
    fft_2d = np.fft.fft2(signal_2d)
    show(np.log1p(np.abs(np.fft.fftshift(fft_2d))), "2D FFT magnitude")
else:
    print("No usable tracking points. Adjust the green mask or video.")
````

**Expected output:** Row/column signals, separate 1D FFT magnitudes, and a 2D FFT magnitude plot. The simple tracker must be checked for missed or incorrect detections.

## 13. Histogram equalization before string-art reconstruction

Save the original image as `stringart.jpg`. This code equalizes the histogram manually using a cumulative distribution lookup table. Then upload both images to https://stringartgenerator.cc/ and compare the string-art results.

````python
img = np.array(Image.open("stringart.jpg").convert("L"))
hist = np.bincount(img.ravel(), minlength=256)
cdf = hist.cumsum()
nonzero = cdf[cdf > 0]
cdf_min = nonzero[0]

if img.size == cdf_min:
    equalized = img.copy()
else:
    lookup = np.clip((cdf - cdf_min) * 255 / (img.size - cdf_min), 0, 255).astype(np.uint8)
    equalized = lookup[img]

Image.fromarray(equalized).save("stringart_equalized.jpg")

plt.subplot(2,2,1); plt.imshow(img, cmap="gray"); plt.title("Original"); plt.axis("off")
plt.subplot(2,2,2); plt.imshow(equalized, cmap="gray"); plt.title("Equalized"); plt.axis("off")
plt.subplot(2,2,3); plt.plot(hist); plt.title("Original histogram")
plt.subplot(2,2,4); plt.hist(equalized.ravel(), bins=256, range=(0,255)); plt.title("Equalized histogram")
plt.tight_layout(); plt.show()

print("Saved equalized image as stringart_equalized.jpg")
````

**How to compare:** Upload the original and equalized images separately to the string-art site. Compare whether important edges and details are more visible. Histogram equalization can improve contrast, but does not always improve string-art quality.
